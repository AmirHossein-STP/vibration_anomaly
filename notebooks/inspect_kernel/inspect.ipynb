```python
import scipy.io, pathlib, numpy as np, glob

mats = glob.glob('/kaggle/input/**/*.mat', recursive=True)
print(f'{len(mats)} .mat files found')
samples = [m for m in mats if 'Normal' in m][:1] + [m for m in mats if 'B007' in m][:1] + [m for m in mats if 'OR007' in m][:1]
for p in samples:
    print('===', p, '===')
    d = scipy.io.loadmat(p)
    for k, v in d.items():
        if k.startswith('__'):
            continue
        arr = np.asarray(v)
        print(f'  {k}: shape={arr.shape} dtype={arr.dtype}')
    print()
```
